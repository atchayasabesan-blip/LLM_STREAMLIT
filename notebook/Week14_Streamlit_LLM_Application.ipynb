```python
import ast

APP_FILE = PROJECT_DIR / "app.py"

with open(APP_FILE, "r", encoding="utf-8") as file:
    app_code = file.read()

tree = ast.parse(app_code)

print("Application File:", APP_FILE.name)
print("Total Lines:", len(app_code.splitlines()))

print("\nImported Modules:")
for node in ast.walk(tree):
    if isinstance(node, (ast.Import, ast.ImportFrom)):
        print(ast.unparse(node))

print("\nFunctions:")
for node in ast.walk(tree):
    if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)):
        print(node.name)
```

```python
import csv

RESULT_FILE = DATASET_DIR / "comparison_results.csv"

if RESULT_FILE.exists():
    with open(RESULT_FILE, "r", encoding="utf-8-sig", newline="") as file:
        reader = csv.DictReader(file)
        rows = list(reader)

    print("COMPARISON RESULTS")
    print("=" * 50)
    print("Total Questions:", len(rows))
    print("Columns:", reader.fieldnames)

    for row in rows[:3]:
        print("\n", row)

else:
    print("Comparison results file not found:", RESULT_FILE)
```
